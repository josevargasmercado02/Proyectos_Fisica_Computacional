# Propiedades ópticas y electrónicas de la plata a partir de datos experimentales

**Objetivo:** extraer $\omega_p$, $\tau$ y $N$ de la plata ajustando el modelo de Drude a los datos de Johnson y Christy (1972), y predecir sus propiedades ópticas macroscópicas ($R$, $\alpha$, $T$).

---

## Marco teórico

### 1. Índice de refracción complejo y función dieléctrica
La respuesta óptica de un material se resume en el índice de refracción complejo

$$\tilde n(\omega)=n(\omega)+ik(\omega)$$

- $n$: propagación de la onda dentro del material.
- $k$: coeficiente de extinción, asociado a la absorción.

Se relaciona con la función dieléctrica $\epsilon(\omega)=\epsilon_1+i\epsilon_2$ mediante $\epsilon=\tilde n^2$. Al desarrollar el cuadrado:

$${\epsilon_1=n^2-k^2,\qquad \epsilon_2=2nk}$$

Esta es la transformación que permite pasar de los datos medidos ($n,k$) a la cantidad que predice la teoría ($\epsilon$). La relación inversa es

$$n=\sqrt{\frac{|\epsilon|+\epsilon_1}{2}},\qquad k=\sqrt{\frac{|\epsilon|-\epsilon_1}{2}}$$

### 2. Modelo de Drude
En un metal, los electrones de conducción se tratan como un gas de electrones libres que se acelera con el campo y pierde momento por colisiones cada tiempo $\tau$:

$$m^*\frac{d\mathbf v}{dt}=-e\mathbf E-\frac{m^*}{\tau}\mathbf v$$

Con un campo armónico $\mathbf E\propto e^{-i\omega t}$ se obtiene la conductividad $\sigma(\omega)=\dfrac{\sigma_0}{1-i\omega\tau}$ y, a partir de ella, la función dieléctrica

$$\boxed{\epsilon(\omega)=\epsilon_\infty-\frac{\omega_p^2}{\omega^2+i\gamma\omega}},\qquad \gamma=\frac1\tau$$

Separando parte real e imaginaria:

$$\epsilon_1(\omega)=\epsilon_\infty-\frac{\omega_p^2}{\omega^2+\gamma^2},\qquad
\epsilon_2(\omega)=\frac{\omega_p^2\,\gamma}{\omega(\omega^2+\gamma^2)}$$

| Parámetro | Significado físico |
|---|---|
| $\omega_p$ | Frecuencia de plasma: oscilación colectiva del gas de electrones |
| $\gamma=1/\tau$ | Amortiguamiento: tasa de colisiones (dispersión) |
| $\epsilon_\infty$ | Fondo dieléctrico de los electrones ligados a altas frecuencias |

### 3. Frecuencia de plasma y concentración de portadores

$$\omega_p^2=\frac{Ne^2}{\epsilon_0 m^*}\quad\Longrightarrow\quad N=\frac{\epsilon_0 m^*\omega_p^2}{e^2}$$

Aquí se asume $m^*\approx m_e$. Por eso $N$ se puede comparar directamente con el valor de Ashcroft–Mermin para la plata ($5.86\times10^{28}\ \text{m}^{-3}$).

### 4. Borde de plasma
Para $\omega\gg\gamma$, $\epsilon_1$ cambia de signo en

$$\omega_{0}\approx\frac{\omega_p}{\sqrt{\epsilon_\infty}}$$

- Para $\omega<\omega_0$: $\epsilon_1<0$, la onda no se propaga y el metal **refleja** ($R\to1$).
- Para $\omega>\omega_0$: $\epsilon_1>0$, el metal se vuelve parcialmente **transparente**.

Con los valores que se obtienen del ajuste ($\omega_p\approx1.39\times10^{16}$ rad/s, $\epsilon_\infty\approx3.58$), el borde cae en $\lambda\approx0.26\ \mu$m, que coincide con la caída de $R$ que se verá en las gráficas.

### 5. Propiedades ópticas macroscópicas (incidencia normal)

$$R=\left|\frac{\tilde n-1}{\tilde n+1}\right|^2,\qquad
\alpha=\frac{4\pi k}{\lambda},\qquad
T=(1-R)\,e^{-\alpha d}$$

La expresión de $T$ para una película de espesor $d$ desprecia las reflexiones múltiples (aproximación simple).



## Datos experimentales
Se descargan los datos de Johnson y Christy (*refractiveindex.info*) y se extraen $\lambda$ (en $\mu$m), $n$ y $k$. Las gráficas de $n(\lambda)$ y $k(\lambda)$ sirven para verificar que los datos tienen el comportamiento metálico esperado, con $k$ creciendo con $\lambda$.

```python
import requests
import yaml
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Descarga del archivo experimental Ag.yml
# Base de datos refractiveindex.info
# Datos de Johnson y Christy para plata
# ============================================================
url = "https://refractiveindex.info/database/data/main/Ag/nk/Johnson.yml"

file_name = "Ag.yml"
print("Descargando archivo experimental...")

response = requests.get(url)

if response.status_code == 200:
    with open(file_name, "wb") as file:
        file.write(response.content)
    print("Archivo descargado correctamente:", file_name)
else:
    raise Exception("No fue posible descargar el archivo")

# ============================================================
# Lectura del archivo YAML
# ============================================================

with open(file_name, "r") as file:
    data = yaml.safe_load(file)

print("\nContenido principal del archivo:")
print(data.keys())

# ============================================================
# Extraccion de datos experimentales
# ============================================================

# Mostrar estructura DATA para verificar formato
print("\nEstructura DATA:")
print(data["DATA"])

raw_data = data["DATA"][0]["data"]

wavelength = []
n_values = []
k_values = []

for line in raw_data.split("\n"):
    if line.strip():
        values = line.split()
        wavelength.append(float(values[0]))
        n_values.append(float(values[1]))
        k_values.append(float(values[2]))

wavelength = np.array(wavelength)
n_values = np.array(n_values)
k_values = np.array(k_values)

print("\nNumero de puntos experimentales:")
print(len(wavelength))

# ============================================================
# Guardar datos procesados
# ============================================================
output = np.column_stack((wavelength, n_values, k_values))

np.savetxt(
    "Ag_optical_data.dat",
    output,
    header="lambda (um) n(lambda) k(lambda)",
    fmt="%.8e"
)

print("\nArchivo generado: Ag_optical_data.dat")

# ============================================================
# Grafica del indice de refraccion
# ============================================================
plt.figure(figsize=(7, 4))

plt.plot(
    wavelength,
    n_values,
    marker="o",
    markersize=3
)

plt.xlabel(r"$\lambda$ ($\mu$m)")
plt.ylabel(r"$n(\lambda)$")
plt.title("Indice de refraccion de la plata")

plt.grid()
plt.tight_layout()

plt.show()

# ============================================================
# Grafica del coeficiente de extincion
# ============================================================
plt.figure(figsize=(7, 4))

plt.plot(
    wavelength,
    k_values,
    marker="o",
    markersize=3
)

plt.xlabel(r"$\lambda$ ($\mu$m)")
plt.ylabel(r"$k(\lambda)$")
plt.title("Coeficiente de extincion de la plata")

plt.grid()
plt.tight_layout()

plt.show()
```

```python
# Conversión de variables experimentales
c = 299792458.0  # Velocidad de la luz en el vacío (m/s)
wavelength_m = wavelength * 1e-6  # Convertimos de micras a metros
omega = (2 * np.pi * c) / wavelength_m  # Frecuencia angular en rad/s

# Cálculo de la función dieléctrica compleja
# eps(w) = eps1(w) + i*eps2(w)
eps1 = n_values**2 - k_values**2
eps2 = 2 * n_values * k_values
```

## Ajuste simultáneo de Drude
Se ajustan $\omega_p$, $\gamma$ y $\epsilon_\infty$ a los datos con `scipy.optimize.curve_fit` (mínimos cuadrados no lineales, Levenberg–Marquardt).

**Punto central:** $\epsilon_1$ y $\epsilon_2$ son dos manifestaciones del mismo material. Ajustarlas por separado daría parámetros distintos, sin sentido físico. Por eso la función `modelo` devuelve ambas **concatenadas** y se ajusta contra `[eps1, eps2]`, de modo que un único conjunto de parámetros minimice el error global.

$$\min_{\omega_p,\gamma,\epsilon_\infty}\ \sum_j\Big[\big(\epsilon_{1,j}-\epsilon_1^{Drude}\big)^2+\big(\epsilon_{2,j}-\epsilon_2^{Drude}\big)^2\Big]$$

Los valores iniciales provienen de la literatura para la plata. Después se grafica el modelo ajustado sobre los datos.

```python
import scipy.optimize as OPT
def modelo(omega, omega_p, gamma, epsilon_inf):
    o_re = epsilon_inf - (omega_p**2)/(omega**2+gamma**2)
    o_im = ((omega_p**2) * gamma)/(omega * (omega**2+gamma**2))
    return np.concatenate((o_re, o_im))

# Concatenamos los datos experimentales
# Se concatenan para que curve_fit optimice todo a la vez
eps_exp_combinado = np.concatenate((eps1, eps2))

par_inicial = [1.38e16, 2.73e13, 3.7] # Valores iniciales para omega_p = 1.38*10^{16}, gamma = 2.73*10^{13}, epsiln_inf = 3.7

popt, pcov = OPT.curve_fit(modelo, omega, eps_exp_combinado, p0=par_inicial)

# Extraemos los datos óptimos del ajuste
omega_p_opt, gamma_opt, epsilon_inf_opt = popt

print("omega_p_opt, gamma_opt, epsilon_inf_opt", popt)

omega_1 = np.linspace(972458217505861.2, 1e16, 5000)

# Escribimos las ecuaciones DIRECTAMENTE aquí, sin llamar a funciones extra
y_1 = epsilon_inf_opt - (omega_p_opt**2)/(omega_1**2 + gamma_opt**2)
y_2 = ((omega_p_opt**2) * gamma_opt)/(omega_1 * (omega_1**2 + gamma_opt**2))

# Gráfica de la parte real
plt.figure(figsize=(7,4))
plt.plot(omega, eps1, marker="o", label="Datos experimentales")
plt.plot(omega_1, y_1, label="Modelo real")
plt.xlabel("Omega [rad/s]")
plt.ylabel("Epsilon 1")
plt.title("Componente real de la funcion dielectrica")
plt.grid()
plt.legend()
plt.tight_layout()
plt.show()

# Gráfica de la parte imaginaria
plt.figure(figsize=(7,4))
plt.plot(omega, eps2, marker="o", label="Datos experimentales")
plt.plot(omega_1, y_2, label="Modelo imaginario")
plt.xlabel("Omega [rad/s]")
plt.ylabel("Epsilon 2")
plt.title("Componente imaginaria de la funcion dielectrica")
plt.grid()
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Extracción de propiedades del material ===
# Extraemos los parámetros de la parte real (omega_p, gamma, epsilon_inf)
omega_p = popt[0]
gamma = popt[1]

# Constantes físicas
e_charge = 1.60217663e-19 # Carga del electrón en C
eps_0 = 8.85418782e-12    # Permitividad del vacío en F/m
m_e = 9.1093837e-31       # Masa del electrón en kg (asumiendo m* aprox m_e)

# Cálculo del tiempo de relajación electrónica
tau = 1.0 / gamma
print(f"Tiempo de relajación electrónica (tau): {tau:.4e} s")

# Cálculo de la concentración de portadores libres (N)
# Despejando N de: omega_p^2 = (N * e^2) / (eps_0 * m_e)
N = (omega_p**2 * eps_0 * m_e) / (e_charge**2)
print(f"Concentración de portadores libres (N): {N:.4e} electrones/m^3")


# === PASO 6: Predicción de propiedades ópticas ===
# 1. Construir el índice de refracción complejo
n_complex = n_values + 1j * k_values

# 2. Coeficiente de reflexión (R)
R = np.abs((n_complex - 1) / (n_complex + 1))**2

# 3. Coeficiente de absorción (alpha)
# Nota: usamos wavelength_m que ya convertiste a metros en el paso anterior
alpha = (4 * np.pi * k_values) / wavelength_m

# 4. Transmisión (T) para un espesor definido 'd'
d = 50e-9 # Ejemplo: espesor de 50 nanómetros
T = (1 - R) * np.exp(-alpha * d)

# === Gráficas de las propiedades ópticas ===
plt.figure(figsize=(15, 4))

# Gráfica del Coeficiente de Reflexión
plt.subplot(1, 3, 1)
plt.plot(wavelength, R, color='blue', marker='.')
plt.xlabel(r'$\lambda$ ($\mu$m)')
plt.ylabel('Reflexión (R)')
plt.title('Coeficiente de Reflexión')
plt.grid(True)

# Gráfica del Coeficiente de Absorción
plt.subplot(1, 3, 2)
plt.plot(wavelength, alpha, color='red', marker='.')
plt.xlabel(r'$\lambda$ ($\mu$m)')
plt.ylabel(r'Absorción $\alpha$ ($m^{-1}$)')
plt.title('Coeficiente de Absorción')
plt.grid(True)

# Gráfica de Transmisión
plt.subplot(1, 3, 3)
plt.plot(wavelength, T, color='green', marker='.')
plt.xlabel(r'$\lambda$ ($\mu$m)')
plt.ylabel('Transmisión (T)')
plt.title(f'Transmisión (espesor d = {d*1e9} nm)')
plt.grid(True)

plt.tight_layout()
plt.show()
```

## Validación Drude vs. experimento
Se reconstruye la respuesta teórica con los parámetros óptimos:

$$\epsilon^{teo}=\epsilon_1^{teo}+i\epsilon_2^{teo}\ \Rightarrow\ \tilde n^{teo}=\sqrt{\epsilon^{teo}}\ \Rightarrow\ n^{teo},\,k^{teo},\,R^{teo},\,\alpha^{teo}$$

y se superpone con los datos en cuatro gráficas ($\epsilon$, $n$ y $k$, $R$, $\alpha$).

```python
# Validación del modelo físico

# 1. Calculamos la Función Dieléctrica TEÓRICA usando los parámetros optimizados
omega_p = popt[0]
gamma = popt[1]
epsilon_inf = popt[2]

eps1_teo = epsilon_inf - (omega_p**2) / (omega**2 + gamma**2)
eps2_teo = (omega_p**2 * gamma) / (omega * (omega**2 + gamma**2))

# 2. Calculamos el Índice de Refracción Complejo TEÓRICO (n + ik)
# Recordando que epsilon = (n + ik)^2, entonces n_complejo = sqrt(epsilon)
eps_complex_teo = eps1_teo + 1j * eps2_teo
n_complex_teo = np.sqrt(eps_complex_teo)

n_teo = n_complex_teo.real
k_teo = n_complex_teo.imag

# 3. Calculamos Propiedades Ópticas TEÓRICAS derivadas
R_teo = np.abs((n_complex_teo - 1) / (n_complex_teo + 1))**2
alpha_teo = (4 * np.pi * k_teo) / wavelength_m

# === GRÁFICAS DE VALIDACIÓN SUPERPUESTAS ===
# Creamos una figura con 4 subgráficos para comparar todo

fig, axs = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Validación del Modelo Físico: Drude vs Datos Experimentales', fontsize=16)

# Gráfica 1: Función Dieléctrica (Real e Imaginaria)
axs[0, 0].plot(omega, eps1, 'bo', markersize=4, label='$\epsilon_1$ Exp')
axs[0, 0].plot(omega, eps1_teo, 'b-', label='$\epsilon_1$ Drude')
axs[0, 0].plot(omega, eps2, 'ro', markersize=4, label='$\epsilon_2$ Exp')
axs[0, 0].plot(omega, eps2_teo, 'r-', label='$\epsilon_2$ Drude')
axs[0, 0].set_xlabel('Frecuencia Angular $\omega$ (rad/s)')
axs[0, 0].set_ylabel('$\epsilon(\omega)$')
axs[0, 0].set_title('Función Dieléctrica')
axs[0, 0].legend()
axs[0, 0].grid(True)

# Gráfica 2: Índice de Refracción (n y k)
axs[0, 1].plot(wavelength, n_values, 'bo', markersize=4, label='n Exp')
axs[0, 1].plot(wavelength, n_teo, 'b-', label='n Drude')
axs[0, 1].plot(wavelength, k_values, 'ro', markersize=4, label='k Exp')
axs[0, 1].plot(wavelength, k_teo, 'r-', label='k Drude')
axs[0, 1].set_xlabel('$\lambda$ ($\mu$m)')
axs[0, 1].set_ylabel('Constantes Ópticas (n, k)')
axs[0, 1].set_title('Índice de Refracción Complejo')
axs[0, 1].legend()
axs[0, 1].grid(True)

# Gráfica 3: Coeficiente de Reflexión (R)
axs[1, 0].plot(wavelength, R, 'go', markersize=4, label='R Exp')
axs[1, 0].plot(wavelength, R_teo, 'g-', label='R Drude')
axs[1, 0].set_xlabel('$\lambda$ ($\mu$m)')
axs[1, 0].set_ylabel('R')
axs[1, 0].set_title('Coeficiente de Reflexión')
axs[1, 0].legend()
axs[1, 0].grid(True)

# Gráfica 4: Coeficiente de Absorción (alpha)
axs[1, 1].plot(wavelength, alpha, 'mo', markersize=4, label=r'$\alpha$ Exp')
axs[1, 1].plot(wavelength, alpha_teo, 'm-', label=r'$\alpha$ Drude')
axs[1, 1].set_xlabel('$\lambda$ ($\mu$m)')
axs[1, 1].set_ylabel(r'$\alpha$ ($m^{-1}$)')
axs[1, 1].set_title('Coeficiente de Absorción')
axs[1, 1].legend()
axs[1, 1].grid(True)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
# Cálculo del Error Absoluto
error_abs_eps1 = np.abs(eps1 - eps1_teo)
error_abs_eps2 = np.abs(eps2 - eps2_teo)

# Cálculo del Error Relativo (evitando división por cero)
error_rel_eps1 = np.where(eps1 != 0, error_abs_eps1 / np.abs(eps1), 0)
error_rel_eps2 = np.where(eps2 != 0, error_abs_eps2 / np.abs(eps2), 0)

# Error Cuadrático Medio (MSE) para medir la desviación global del modelo
mse_eps1 = np.mean((eps1 - eps1_teo)**2)
mse_eps2 = np.mean((eps2 - eps2_teo)**2)

print("--- Análisis de Error Numérico ---")
print(f"MSE Epsilon 1 (Parte Real): {mse_eps1:.4e}")
print(f"MSE Epsilon 2 (Parte Imag): {mse_eps2:.4e}")
print(f"Error Relativo Promedio Epsilon 1: {np.mean(error_rel_eps1)*100:.2f}%")
print(f"Error Relativo Promedio Epsilon 2: {np.mean(error_rel_eps2)*100:.2f}%")
```

```python
print("\n--- Análisis de Convergencia e Incertidumbre ---")
# La matriz pcov permite calcular el error estándar de los parámetros ajustados
incertidumbre_params = np.sqrt(np.diag(pcov))

print(f"Parámetros óptimos [omega_p, gamma, eps_inf]: {popt}")
print(f"Incertidumbre (desviación) del ajuste numérico: {incertidumbre_params}")

print("\nPrueba con diferentes valores iniciales (semillas):")
# Definimos valores iniciales extremos para ver si el algoritmo se pierde o si logra llegar al mismo resultado
semillas_prueba = [
    [1.38e16, 2.73e13, 3.7],    # Semilla base
    [1.0e15, 1.0e12, 1.0],      # Semilla subestimada
    [5.0e16, 5.0e14, 10.0]      # Semilla sobrestimada
]

for i, p0_test in enumerate(semillas_prueba):
    try:
        # Aumentamos maxfev por si necesita más iteraciones desde un punto lejano
        popt_test, _ = OPT.curve_fit(modelo, omega, eps_exp_combinado, p0=p0_test, maxfev=10000)

        # Comparamos si llegó a los mismos parámetros
        print(f"Semilla {i+1} {p0_test}: \n  Convergió a -> {popt_test}")
    except Exception as e:
        print(f"Semilla {i+1} {p0_test}: \n  Falla de convergencia. Motivo: {e}")
```

### Límite de validez del modelo
Drude solo considera electrones **libres**. En la plata, a energías del visible y ultravioleta los fotones excitan **transiciones interbanda** (banda $d$), un efecto que el modelo no contiene y que $\epsilon_\infty$ solo compensa parcialmente. Se espera, entonces, buen ajuste en el infrarrojo y desviación sistemática a longitud de onda corta. La solución natural es Drude–Lorentz.
